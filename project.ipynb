```python
# Step 1: Define sample data
customer_names = ["Alice", "Bob", "Charlie", "Diana", "Eve", "Frank"]

orders = [
    ("Alice", "Laptop", 1200, "Electronics"),
    ("Alice", "Mouse", 25, "Electronics"),
    ("Bob", "T-Shirt", 30, "Clothing"),
    ("Bob", "Jeans", 50, "Clothing"),
    ("Charlie", "Blender", 80, "Home Essentials"),
    ("Diana", "Headphones", 150, "Electronics"),
    ("Diana", "Desk Lamp", 45, "Home Essentials"),
    ("Eve", "Sweater", 60, "Clothing"),
    ("Eve", "Socks", 15, "Clothing"),
    ("Frank", "Coffee Maker", 120, "Home Essentials"),
    ("Frank", "Toaster", 65, "Home Essentials"),
]

customer_products = {
    "Alice": ["Laptop", "Mouse"],
    "Bob": ["T-Shirt", "Jeans"],
    "Charlie": ["Blender"],
    "Diana": ["Headphones", "Desk Lamp"],
    "Eve": ["Sweater", "Socks"],
    "Frank": ["Coffee Maker", "Toaster"]
}

# Step 2: Product-to-category mapping
product_category = {
    "Laptop": "Electronics", "Mouse": "Electronics",
    "T-Shirt": "Clothing", "Jeans": "Clothing",
    "Blender": "Home Essentials",
    "Headphones": "Electronics", "Desk Lamp": "Home Essentials",
    "Sweater": "Clothing", "Socks": "Clothing",
    "Coffee Maker": "Home Essentials", "Toaster": "Home Essentials"
}

unique_categories = set(product_category.values())

# Step 3: Calculate customer spending
customer_spending = {}
for customer, product, price, category in orders:
    if customer not in customer_spending:
        customer_spending[customer] = 0
    customer_spending[customer] += price

# Step 4: Classify customers
customer_classification = {}
for customer, total_spending in customer_spending.items():
    if total_spending > 100:
        classification = "High-Value Buyer"
    elif total_spending >= 50:
        classification = "Moderate Buyer"
    else:
        classification = "Low-Value Buyer"
    customer_classification[customer] = {
        "total_spending": total_spending,
        "classification": classification
    }

# Step 5: Top 3 customers
top_customers = sorted(customer_spending.items(), 
                       key=lambda x: x[1], reverse=True)[:3]

# Step 6: Revenue per category
category_revenue = {}
for customer, product, price, category in orders:
    if category not in category_revenue:
        category_revenue[category] = 0
    category_revenue[category] += price

# Step 7: Unique products
unique_products = set(product for _, product, _, _ in orders)

# Step 8: Electronics customers
electronics_customers = set(
    customer for customer, _, _, category in orders if category == "Electronics"
)

# Step 9: Multi-category customers
customer_categories = {}
for customer, product, price, category in orders:
    if customer not in customer_categories:
        customer_categories[customer] = set()
    customer_categories[customer].add(category)

multi_category_customers = {
    customer: categories for customer, categories in customer_categories.items()
    if len(categories) > 1
}

# Step 10: Both Electronics AND Clothing
electronics_buyers = set(
    customer for customer, _, _, category in orders if category == "Electronics"
)
clothing_buyers = set(
    customer for customer, _, _, category in orders if category == "Clothing"
)
both_categories = electronics_buyers & clothing_buyers

# Step 11: Print comprehensive report
print("\n" + "="*70)
print("COMPREHENSIVE CUSTOMER ORDER ANALYSIS REPORT".center(70))
print("="*70)

print(f"\n📊 KEY METRICS")
print(f"  Total Customers: {len(customer_names)}")
print(f"  Total Orders: {len(orders)}")
print(f"  Total Revenue: ${sum(category_revenue.values())}")

print(f"\n📂 PRODUCT CATEGORIES")
print(f"  Unique Categories: {sorted(unique_categories)}")
print(f"  Unique Products: {len(unique_products)}")

print(f"\n💰 CUSTOMER SPENDING & CLASSIFICATION")
for customer in sorted(customer_classification.keys()):
    info = customer_classification[customer]
    print(f"  {customer}: ${info['total_spending']} - {info['classification']}")

print(f"\n🏆 TOP 3 HIGHEST-SPENDING CUSTOMERS")
for rank, (customer, spending) in enumerate(top_customers, 1):
    print(f"  {rank}. {customer}: ${spending}")

print(f"\n📈 REVENUE BY CATEGORY")
for category, revenue in sorted(category_revenue.items()):
    pct = (revenue / sum(category_revenue.values())) * 100
    print(f"  {category}: ${revenue} ({pct:.1f}%)")

print(f"\n🛒 ELECTRONICS CUSTOMERS")
for customer in sorted(electronics_customers):
    print(f"  • {customer}")

print(f"\n🔄 MULTI-CATEGORY BUYERS")
for customer in sorted(multi_category_customers.keys()):
    print(f"  {customer}: {sorted(multi_category_customers[customer])}")

print(f"\n✨ CUSTOMERS WHO BOUGHT BOTH ELECTRONICS & CLOTHING")
for customer in sorted(both_categories):
    print(f"  • {customer}")
if not both_categories:
    print("  None")

print("\n" + "="*70)
```
